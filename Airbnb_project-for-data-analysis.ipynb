```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Airbnb_data.csv")
```

```python
df
```

```python
df.columns
```

# checking  missing values 

```python
print(df.isnull().sum())
```

```python
df.info()
```

```python
df['last review']=pd.to_datetime(df["last review"], errors='coerce')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.fillna({'number of reviews':0,'last review':df['last review'].min()},inplace=True)
```

```python
df.dropna(subset=['NAME','host id'],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.drop(columns=['license','house_rules'],errors='ignore')
```

```python
df.head()
```

# Remove dollor signs and converts float

```python
df['price']=df['price'].replace("[\$,]","",regex=True).astype(float)
```

```python
df['service fee']=df['service fee'].replace("['\$']",'', regex=True).astype(float)
```

```python
df.head(2)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.describe()
```

# visulization

# What is distribution of listing price ?


```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],bins=50,kde=True,color='red')
plt.title("Distribution of price ")
plt.xlabel('Price $')
plt.ylabel('Count')
```

# how are the different type of room distributed ?

```python
plt.figure(figsize=(6,4))
sns.countplot(x='room type',data=df)
```

# How  are listings distributed across different neighborhoods?

```python
plt.figure(figsize=(10,8))
sns.countplot(y='neighbourhood group',data=df , color='lightgreen', order=df['neighbourhood group'].value_counts().index)
plt.title("number of listings by neighborhood")
plt.xlabel("Count")
plt.ylabel('Neighborhood group')
plt.show()
```

# what is the relationship between peice and room  ?

```python
plt.figure(figsize=(10,8))
sns.boxplot(x='room type', y='price',hue='room type',data=df,palette='Set1')
```
